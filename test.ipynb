```python
from pandas import (
    DataFrame,
    read_excel,
    Series,
)
import matplotlib.pyplot as plt
```

```python
results = read_excel('.\\tests\\rag\\results\\rag_results.xlsx', index_col=0)
results
```

```python
def key(name: str) -> str:
    return f'{name[:8]}, {name[15:]}'
```

```python
aggregated = results.groupby(key).aggregate('mean').sort_values(by='nv_accuracy').reset_index()
aggregated
```

```python
keys: list[str] = [
    'baseline, 25_12-33',
    '22-05-25, without-filter',
    '20-05-25, ',
    '21-05-25, ',
    '03-06-25, with-keyword',
]
```

```python
xs = [
    'baseline',
    'basic',
    'filtered',
    'improved html',
    'final',
]
```

```python
filtered = aggregated[[name in keys for name in aggregated['name']]]
```

```python
fig, ax = plt.subplots(1, 1, figsize=(10, 5))
ax.plot(xs, filtered['nv_accuracy'], label='Accuracy')
ax.plot(xs, filtered['nv_context_relevance'], label='Context Relevance')
ax.plot(xs, filtered['nv_response_groundedness'], label='Response Groundedness')
ax.legend()
ax.set_ylim(0, 1)
ax.set_yticks([i/10 for i in range(11)])
ax.grid()

plt.savefig('test.png')
```
